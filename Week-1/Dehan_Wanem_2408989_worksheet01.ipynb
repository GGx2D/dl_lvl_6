```python
# Convert the original image to grayscale
gray_image = org_image.convert("L")

# Convert grayscale image to NumPy array
gray_array = np.array(gray_image, dtype=np.float64)

# Display the grayscale image
plt.imshow(gray_array, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

• Center the dataset

```python
# Calculate mean of each column
mean_vector = np.mean(gray_array, axis=0)

# Subtract mean from each column
centered_data = gray_array - mean_vector

# Display centered data
plt.imshow(centered_data, cmap="gray")
plt.title("Centered Data")
plt.axis("off")
plt.show()

# Print shape
print("Centered data shape:", centered_data.shape)
```

Standarize the dataset

```python
# Calculate standard deviation of each column
std_vector = np.std(centered_data, axis=0)

# Avoid division by zero
std_vector[std_vector == 0] = 1

# Standardize data
standardized_data = centered_data / std_vector

# Display standardized data
plt.imshow(standardized_data, cmap="gray")
plt.title("Standardized Data")
plt.axis("off")
plt.show()

# Print shape
print("Standardized data shape:", standardized_data.shape)
```

• Calculate the covaraince matrix of the Standaridze data.

```python
# Calculate covariance matrix of standardized data
cov_matrix = np.cov(standardized_data, rowvar=False)

# Print covariance matrix shape
print("Covariance matrix shape:", cov_matrix.shape)
```

**2. Eigen Decomposition and Identifying Principal Components:**

• Compute Eigen Values and Eigen Vectors.

```python
# Compute eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

# Print shapes
print("Eigenvalues shape:", eigenvalues.shape)
print("Eigenvectors shape:", eigenvectors.shape)
```

• Sort the eigenvalues in descending order and choose the top k eigenvectors corresponding to
the highest eigenvalues.

```python
# Sort eigenvalues in descending order
sorted_indices = np.argsort(eigenvalues)[::-1]

# Sort eigenvalues
sorted_eigenvalues = eigenvalues[sorted_indices]

# Sort eigenvectors
sorted_eigenvectors = eigenvectors[:, sorted_indices]
```

```python
# Choose top k principal components
k = 50
top_k_eigenvectors = sorted_eigenvectors[:, :k]

# Print shape
print("Top k eigenvectors shape:", top_k_eigenvectors.shape)
```

• Identify the Principal Components with the help of cumulative Sum plot.

```python
# Calculate explained variance ratio
explained_variance_ratio = sorted_eigenvalues / np.sum(sorted_eigenvalues)

# Calculate cumulative explained variance
cumulative_variance = np.cumsum(explained_variance_ratio)

# Plot cumulative explained variance
plt.figure(figsize=(8, 5))
plt.plot(cumulative_variance, marker='o')
plt.title("Cumulative Explained Variance")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.grid(True)
plt.show()
```

**3. Reconstruction and Experiment:**

• Reconstruction:  Transform the original data by multiplying it with the selected eigenvectors(PCs) to obtain a lower-dimensional representation.

```python
# Choose number of principal components
k = 50

# Select top k eigenvectors
selected_eigenvectors = sorted_eigenvectors[:, :k]

# Transform data into lower-dimensional representation
compressed_data = np.dot(standardized_data, selected_eigenvectors)

# Reconstruct the image
reconstructed_data = np.dot(compressed_data, selected_eigenvectors.T)

# Convert back to original scale
reconstructed_data = reconstructed_data * std_vector + mean_vector

# Keep pixel values in valid range
reconstructed_data = np.clip(reconstructed_data, 0, 255)
```

```python
# Display original grayscale cat image and reconstructed cat image
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray_array, cmap="gray")
plt.title("Original Grayscale Butterfly Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(reconstructed_data, cmap="gray")
plt.title(f"Reconstructed Butterfly Image (k={k})")
plt.axis("off")

plt.show()
```

• Experiments: Pick Four different combination of principal components with various explained variance value and compare the result.

```python
# Four different numbers of principal components
k_values = [10, 20, 50, 100]

# Create figure for comparison
plt.figure(figsize=(12, 8))

# Show original grayscale cat image
plt.subplot(2, 3, 1)
plt.imshow(gray_array, cmap="gray")
plt.title("Original Grayscale Butterfly Image")
plt.axis("off")

# Reconstruct image for each k
for i, k in enumerate(k_values):
    # Select top k eigenvectors
    selected_eigenvectors = sorted_eigenvectors[:, :k]

    # Compress the data
    compressed_data = np.dot(standardized_data, selected_eigenvectors)

    # Reconstruct the image
    reconstructed_data = np.dot(compressed_data, selected_eigenvectors.T)

    # Convert back to original scale
    reconstructed_data = reconstructed_data * std_vector + mean_vector

    # Keep values in valid range
    reconstructed_data = np.clip(reconstructed_data, 0, 255)

    # Display result
    plt.subplot(2, 3, i + 2)
    plt.imshow(reconstructed_data, cmap="gray")
    plt.title(f"Reconstructed Butterfly Image (k={k})")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

Experiment using explained variance values

```python
# Four explained variance values
variance_values = [0.80, 0.90, 0.95, 0.99]

# Find number of principal components needed for each explained variance
k_from_variance = []

for value in variance_values:
    k = np.argmax(cumulative_variance >= value) + 1
    k_from_variance.append(k)

# Print result
print("Number of principal components needed:")
for value, k in zip(variance_values, k_from_variance):
    print(f"{int(value*100)}% explained variance -> k = {k}")
```

• Display the Results

```python
# Create figure
plt.figure(figsize=(12, 8))

# Display original image
plt.subplot(2, 3, 1)
plt.imshow(gray_array, cmap="gray")
plt.title("Original Image")
plt.axis("off")

# Reconstruct and display for each explained variance
for i, k in enumerate(k_from_variance):
    # Select PCs
    selected_eigenvectors = sorted_eigenvectors[:, :k]

    # Compress
    compressed_data = np.dot(standardized_data, selected_eigenvectors)

    # Reconstruct
    reconstructed_data = np.dot(compressed_data, selected_eigenvectors.T)

    # Convert back to original scale
    reconstructed_data = reconstructed_data * std_vector + mean_vector

    # Clip values
    reconstructed_data = np.clip(reconstructed_data, 0, 255)

    # Display image
    plt.subplot(2, 3, i + 2)
    plt.imshow(reconstructed_data, cmap="gray")
    plt.title(f"{int(variance_values[i]*100)}% variance\nk={k}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

and Evaluate.

```python
# Evaluate compression ratio for each explained variance
original_size = gray_array.shape[1]   # original number of columns/features

print("Evaluation:")
for value, k in zip(variance_values, k_from_variance):
    compression_ratio = k / original_size
    print(f"{int(value*100)}% variance -> k = {k}, Compression ratio = {compression_ratio:.4f}")
```
